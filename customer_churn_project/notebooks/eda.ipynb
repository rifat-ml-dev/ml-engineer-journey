```python
import pandas as pd

orginal_df = pd.read_csv("../data/telco.csv")
df = orginal_df.copy()
df.head(10)
```

```python
# Exploratory Data Analysis
print("Shape:",df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nInfo:")
print(df.info())

print("\nTarget Distribuion:")
print(df['Churn'].value_counts())
print()
print(df['Churn'].value_counts(normalize=True)*100)
```

```python
print(df.head(5))
df.dtypes
```

```python
df['TotalCharges'] = pd.to_numeric(
    df['TotalCharges'],
    errors='coerce'
)

print(df['TotalCharges'].dtype)
print(df['TotalCharges'].isnull().sum())
```

```python
df.info()
```

```python
df = df.dropna()
print(df.shape)
print(df.isnull().sum().sum())
```

```python
#Feature Analysis
print(df.nunique())
```
